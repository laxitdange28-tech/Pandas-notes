## Q21. Remove everything after '(' from Review.

```python
def remove_text(text):
  if '(' in text:
    return text[:text.index('(')].strip()
```

```python
df['Review'] = df['Review'].apply(remove_text)
df
```

## Q24. Remove extra spaces from Customer.

```python
# Write your code here
def remove_space(Customers):
  return "".join(Customers.split(" "))
```

```python
df['Customer'] = df['Customer'].apply(remove_space)
df
```
