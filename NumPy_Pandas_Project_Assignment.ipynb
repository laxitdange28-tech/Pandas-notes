## Question 1

Create a NumPy array using the salary values:

Input:
```
[50000,42000,65000,58000,45000]
```

Output:
```
array([50000,42000,65000,58000,45000])
```

````python
# Write your code here
import numpy as np

salary = np.array([50000,42000,65000,58000,45000])
print(salary)
````

## Question 2

Using the NumPy array, find Total Salary, Average Salary and Maximum Salary.

Output:
```
Total Salary = 260000
Average Salary = 52000.0
Maximum Salary = 65000
```

````python
# Write your code here
total_salary = np.sum(salary)
average_salary = np.mean(salary)
max_salary = np.max(salary)

print(f"Totall salary {total_salary}")
print(f"Average Slary {average_salary}")
print(f"Max slary {max_salary}")
````

## Question 3

Convert the NumPy array into a Pandas DataFrame.

Column Name: **Salary**

````python
import pandas as pd
df = pd.DataFrame(salary , columns=["Salary"])
df
````
